### Passo 1 - Importar Arquivos e Bibliotecas

```python
senha_app = "pgxv kaxz zgho gtos"
```

```python
import pandas as pd
from pathlib import Path 
from email.message import EmailMessage
import smtplib
caminho_importacao = Path("Bases de Dados")

email_gerente = pd.read_excel(caminho_importacao / "Emails.xlsx")
lojas = pd.read_csv(caminho_importacao / "Lojas.csv", encoding="latin1", sep=";")
vendas = pd.read_excel(caminho_importacao / "Vendas.xlsx")

display(email_gerente)
display(lojas)
vendas = vendas.merge(lojas, on="ID Loja")
display(vendas)
```

### Passo 2 - Definir Criar uma Tabela para cada Loja e Definir o dia do Indicador

```python
dia_indicador = vendas["Data"].max()


dic_lojas = {}

for loja in lojas["Loja"]:
    dic_lojas[loja] = vendas.loc[vendas["Loja"]==loja, :]


gerente = email_gerente.loc[email_gerente["Loja"]=="Iguatemi Esplanada", "Gerente"].values[0]
print(gerente)
```

### Passo 3 - Salvar a planilha na pasta de backup

```python
caminho_backup = Path(r"Backup Arquivos Lojas")
#lista_arquivos = caminho_backup.iterdir()
#lista_nomes = [for arquivo ]


for loja in dic_lojas:
    pastas_lojas = caminho_backup / loja
    pastas_lojas.mkdir(exist_ok=True)

    nome_arquivo = f"{dia_indicador.month}_{dia_indicador.day}_{loja}.xlsx"
    arquivos = pastas_lojas / nome_arquivo
    dic_lojas[loja].to_excel(arquivos, index=False)
    
```

```python
# metas 
meta_faturamento_anual = 1650000
meta_faturamento_dia = 1000

meta_produtos_anual = 120
meta_produtos_dia = 4

meta_ticket_medio_anual = 500
meta_ticket_medio_dia = 500
```

### Passo 4 - Calcular o indicador para 1 loja

```python
email_remetente = "lucasrossi1422@gmail.com"
senha_aplicativo = (
    senha_app  
)



print("Conectando ao servidor de e-mail...")
servidor = smtplib.SMTP("smtp.gmail.com", 587)
servidor.starttls() 
servidor.login(email_remetente, senha_aplicativo)
print("Conexão estabelecida com sucesso!\n")



for loja in dic_lojas:
    vendas_anuais = dic_lojas[loja]
    vendas_dia = vendas_anuais.loc[vendas_anuais["Data"]==dia_indicador, :]

    faturamento_anual = vendas_anuais["Valor Final"].sum() 
    faturamento_dia = vendas_dia["Valor Final"].sum()

    produtos_anual = len(vendas_anuais["Produto"].unique())  
    produtos_dia = len(vendas_dia["Produto"].unique())

    ticket_medio_anual = vendas_anuais[["Código Venda", "Valor Final"]].groupby("Código Venda").sum()
    ticket_medio_anual = ticket_medio_anual["Valor Final"].mean()

    ticket_medio_dia = vendas_dia[["Código Venda", "Valor Final"]].groupby('Código Venda').sum()
    ticket_medio_dia = ticket_medio_dia["Valor Final"].mean()

    if faturamento_anual > meta_faturamento_anual:
        cor_faturamento_anual = "green"
    else:
        cor_faturamento_anual = "red"

    if faturamento_dia > meta_faturamento_dia:
        cor_faturamento_dia = "green"
    else:
        cor_faturamento_dia = "red"

    if produtos_anual > meta_produtos_anual:
        cor_produtos_anual = "green"
    else:
        cor_produtos_anual = "red"

    if produtos_dia > meta_produtos_dia:
        cor_produtos_dia = "green"
    else:
        cor_produtos_dia = "red"

    if ticket_medio_anual > meta_ticket_medio_anual:
        cor_ticket_anual = "green"
    else:
        cor_ticket_anual = "red"

    if ticket_medio_dia > meta_ticket_medio_dia:
        cor_ticket_dia = "green"
    else:
         cor_ticket_dia = "red"


    nome_arquivo = f"{dia_indicador.month}_{dia_indicador.day}_{loja}.xlsx"
    caminho_arquivo = caminho_backup / loja / nome_arquivo


    email_destinatario = email_gerente.loc[email_gerente["Loja"]==loja, "E-mail"].values[0]
    nome_gerente = email_gerente.loc[email_gerente["Loja"]==loja, "Gerente"].values[0]
    
    corpo_email = f"""
    <p>Bom dia, {nome_gerente}</p>
    
    <p>O resultado de ontem ({dia_indicador.strftime('%d/%m/%Y')}) da Loja {loja} foi:</p>
    
    <table>
  <thead>
    <tr>
      <th scope="col">Indicadores</th>
      <th scope="col">Valor dia</th>
      <th scope="col">Meta Dia</th>
      <th scope="col">Cenário Dia</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <th scope="row">Faturamento</th>
      <td>{faturamento_dia:,.2f}</td>
      <td>{meta_faturamento_dia:,.2f}</td>
      <td><span style="color: {cor_faturamento_dia}; font-size: 32px;">●</span></td>
    </tr>
    <tr>
      <th scope="row">Produtos</th>
      <td>{produtos_dia}</td>
      <td>{meta_produtos_dia}</td>
      <td><span style="color: {cor_produtos_dia}; font-size: 32px;">●</span></td>
    </tr>
    <tr>
      <th scope="row">Ticket Médio</th>
      <td>{ticket_medio_dia:,.2f}</td>
      <td>{meta_ticket_medio_dia:,.2f}</td>
      <td><span style="color: {cor_ticket_dia}; font-size: 32px;">●</span></td>
    </tr>
  </tbody>
</table>
    

       <table>
      <thead>
        <tr>
          <th scope="col">Indicadores</th>
          <th scope="col">Valor anual</th>
          <th scope="col">Meta anual</th>
          <th scope="col">Cenário anual</th>
        </tr>
      </thead>
      <tbody>
        <tr>
          <th scope="row">Faturamento</th>
          <td>{faturamento_anual:,.2f}</td>
          <td>{meta_faturamento_anual:,.2f}</td>
          <td><span style="color: {cor_faturamento_anual}; font-size: 32px;">●</span></td>
        </tr>
        <tr>
          <th scope="row">Produtos</th>
          <td>{produtos_anual}</td>
          <td>{meta_produtos_anual}</td>
          <td><span style="color: {cor_produtos_anual}; font-size: 32px;">●</span></td>
        </tr>
        <tr>
          <th scope="row">Ticket Médio</th>
          <td>{ticket_medio_anual:,.2f}</td>
          <td>{meta_ticket_medio_anual:,.2f}</td>
          <td><span style="color: {cor_ticket_anual}; font-size: 32px;">●</span></td>
        </tr>
      </tbody>
    </table>

    <p>Segue em anexo a planilha com todos os dados para mais detalhes.<br>
    Qualquer dúvida, estou à disposição.</p>
    
    <p>Att.,<br>
    Lucas</p>
    """
    email = EmailMessage()
    email["Subject"] = (
      f"OnePage Dia {dia_indicador.strftime('%d/%m/%Y')} - Loja {loja}"
  )
    email["From"] = email_remetente
    email["To"] = email_destinatario
    
      
    email.set_content(corpo_email, subtype="html"
          
      )
    
      
    if caminho_arquivo.exists():
        with open(caminho_arquivo, "rb") as arquivo:
          dados_arquivo = arquivo.read()
          email.add_attachment(
              dados_arquivo,
              maintype="application",
              subtype="octet-stream",
              filename=nome_arquivo,
          )
    
        
        servidor.send_message(email)
        print(f"E-mail enviado com sucesso para a loja: {loja} ({email_destinatario})")
    else:
        print(f"Aviso: Arquivo não encontrado para a loja {loja} em {caminho_arquivo}")
    
    
servidor.quit()
print("\nProcesso de envio finalizado e conexão encerrada.")


```
